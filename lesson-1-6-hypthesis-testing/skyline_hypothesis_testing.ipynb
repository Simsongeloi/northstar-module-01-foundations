# Calculation the Standard deviation (STD)

```python
# Select only the hours_studied column
python_hours = skyline.loc[
    skyline["course_name"] == "Python for Beginners",
    "hours_studied"
]

sql_hours = skyline.loc[
    skyline["course_name"] == "SQL Basics",
    "hours_studied"
]

# Calculate standard deviations
python_std = python_hours.std()
sql_std = sql_hours.std()

# Display results
print("Python standard deviation:", round(python_std, 2))
print("SQL standard deviation:", round(sql_std, 2))
```

# Calcualting the mean difference

```python
mean_difference = python_mean - sql_mean
print("mean difference:", round(mean_difference, 2))
```

# Calculating the T test

```python
t_stat, p_value = stats.ttest_ind(
    python_hours,
    sql_hours,
    equal_var=False

)

print("T-statistic:", round(t_stat, 4))
print("P-value:", round(p_value, 4))
```

1. Understanding your t-statistic
Your t-statistic is −0.2107.

The negative sign indicates that the Python group's sample mean is lower than the SQL group's sample mean, based on the order you passed them to the test.

The small absolute value tells us that the observed difference is small relative to its estimated standard error.

2. Understanding your p-value
Your p-value is 0.835.

A p-value tells us how likely we would be to observe a test statistic at least as extreme as yours, assuming the two population means are equal and the test assumptions hold.

Let's compare your result with the significance threshold.

# Calculate Standard error


```python
se = np.sqrt(
    python_std**2 / python_n
    +sql_std**2 / sql_n

)

print("Standard error:", round(se, 4))
```

# This explain that the two courses, Python for beginners and SQL basics have an average standard error of 2.82 hours. 

# Calculate the confidence interval (CI) for the two courses


```python
df = (
    (python_std**2 / python_n + sql_std**2 / sql_n)**2
    /
    (
        (python_std**2 / python_n)**2 / (python_n - 1)
        +
        (sql_std**2 / sql_n)**2 / (sql_n - 1)
    )
)

print("Degrees of freedom:", round(df, 2))
```

```python
t_critical = stats.t.ppf(0.975, df)

margin_error = t_critical * se

ci_lower = mean_difference - margin_error
ci_upper = mean_difference + margin_error

print("95% Confidence Interval:")
print("Lower bound:", round(ci_lower, 2))
print("Upper bound:", round(ci_upper, 2))
```

# The Effect size

```python
pooled_std = np.sqrt(
    (
        (python_n - 1) * python_std**2
        + (sql_n - 1) * sql_std**2
    )
    / (python_n + sql_n - 2)
)

print("Pooled standard deviation:", round(pooled_std, 2))
```

```python
cohens_d = mean_difference / pooled_std

print("Cohen's d:", round(cohens_d, 3))
```

```python
results = []

for course_a, course_b in course_pairs:

    hours_a = skyline.loc[
        skyline["course_name"] == course_a,
        "hours_studied"
    ]

    hours_b = skyline.loc[
        skyline["course_name"] == course_b,
        "hours_studied"
    ]

    mean_diff = hours_a.mean() - hours_b.mean()

    t_stat, p_value = stats.ttest_ind(
        hours_a,
        hours_b,
        equal_var=False
    )

    results.append({
        "course_a": course_a,
        "course_b": course_b,
        "mean_difference": mean_diff,
        "p_value": p_value
    })
```

```python
results_df = pd.DataFrame(results)

print(results_df.to_string(index=False))
```

```python
print("Number of tests:", len(results_df))
```

```python
alpha = 0.05

significant_tests = results_df[
    results_df["p_value"] < alpha
]

print("Significant tests:", len(significant_tests))
```

```python
number_of_tests = len(results_df)

bonferroni_alpha = alpha / number_of_tests

print("Bonferroni threshold:", bonferroni_alpha)
```

```python
bonferroni_significant = results_df[
    results_df["p_value"] < bonferroni_alpha
]

print("Tests surviving Bonferroni:",
      len(bonferroni_significant))
```
