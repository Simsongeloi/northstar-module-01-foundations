```python
import pandas as pd

skyline = pd.read_csv(
    "../lesson-1-2-types-of-data/skyline_enrollments.csv"
)

skyline.head(50)
```

# Scale Of Measurement For Each Column

## Enrollment_id  is "Nominal
## Course_name is "Nominal"
## Enrollment_year is "Interval"
## Final_grade is "Ordinal"
## Hours_studied is " Ratio"
## Completion_status is " Nominal"


# Nominal Column


```python
total_enrollments_id_count = skyline["enrollment_id"].count()
print(f"Total number of enrollments: {total_enrollments_id_count}")
```

```python
total_enrollment_id_mode = skyline["enrollment_id"].mode()
print(f"Most common enrollment_id: {total_enrollment_id_mode[0]}")
```

```python
total_enrollment_id_unique  = skyline["enrollment_id"].nunique()
print(f"Total number of unique enrollment_id: {total_enrollment_id_unique}")
```

```python
total_course_name = skyline["course_name"].nunique()
print(f"Total number of course_name: {total_course_name}")
```

```python
total_course_name_mode = skyline["course_name"].mode()
print(f"Most common course_name: {total_course_name_mode[0]}")
```

```python
total_completion_status_mode = skyline["completion_status"].mode()
print(f"Most common completion_status: {total_completion_status_mode[0]}")
```

```python
total_completed = (skyline["completion_status"] == "completed").sum()

print("Total completed:", total_completed)
```

# Interval Column

```python
total_enrollment_year_mean = skyline["enrollment_year"].mean()
print(f"Mean enrollment_year: {total_enrollment_year_mean}")
```

```python
total_enrollment_year_median = skyline["enrollment_year"].median()
print(f"Median enrollment_year: {total_enrollment_year_median}")
```

```python
total_enrollment_year_mode = skyline["enrollment_year"].mode()
print(f"Most common enrollment_year: {total_enrollment_year_mode[0]}")
```

```python
std_enrollment_year = skyline["enrollment_year"].std()

print("Standard deviation of enrollment year:", round(std_enrollment_year, 2))
```

# Ration Column


```python
total_hours_studied_mean = skyline["hours_studied"].mean()

print("Average hours studied:", round(total_hours_studied_mean, 2))
```

```python
total_hours_studied_median = skyline["hours_studied"].median()
print("Median hours studied:", total_hours_studied_median)
```

```python
total_hours_studied_mode = skyline["hours_studied"].mode()
print("Most common hours studied:", total_hours_studied_mode[0])
```

```python
std_hours_studied = skyline["hours_studied"].std()
print("Standard deviation of hours studied:", round(std_hours_studied, 2))
```

```python
Q1 = skyline["enrollment_year"].quantile(0.25)
print("first quartile:", Q1)

Q3 = skyline["enrollment_year"].quantile(0.75)

print("third quartile:", Q3)
IQR = Q3 - Q1
print(Q1, Q3, IQR)
```

```python
total_hours_studied_p95 = skyline["hours_studied"].quantile(0.95)

print("P95 Hours Studied:", round(total_hours_studied_p95, 2))
```
