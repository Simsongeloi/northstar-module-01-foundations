# Confounder Hypothesis
Althought it seems that hours_studied and final_grade could have a correlation, there are other factors that can be a big potential in final grade of a student. Things like; attencdance, prior knowledge, class participation, access to resources, discipline, etc..

```python

from statsmodels.stats.power import NormalIndPower

from statsmodels.stats.proportion import proportion_effectsize


# Knowing the baseline and target success

baseline = 0.40
target = 0.45

# To calculate the power analysis
effect_size = proportion_effectsize(baseline, target)


analysis = NormalIndPower()

# calculation of the sample size.
sample_size = analysis.solve_power(
    effect_size=effect_size,
    alpha=0.05,
    ratio=1.0,
    power=0.80,
    alternative="two-sided"
)

n_per_group = int(np.ceil(sample_size))


print("Student per group:", n_per_group)
print("Total students:", n_per_group * 2)
```
